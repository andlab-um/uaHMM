```python
import os
import nibabel as nib
import numpy as np
import pandas as pd

data_path = 'subj_Data' 
output_excel = 'activation_data.xlsx'
lie_indices = [1, 4, 7, 10, 13, 16, 19, 22]
honesty_indices = [2, 5, 8, 11, 14, 17, 20, 23]
random_indices = [3, 6, 9, 12, 15, 18, 21, 24]

all_subjects_elie = []
all_subjects_ehon = []
all_subjects_erand = []
all_subjects_eall = []


subject_ids = range(101, 146) # this will skip some participant. No worries

for subject_id in subject_ids:
    subject_dir = os.path.join(data_path, f'sub-{subject_id}')
    
    subj_lie_data = []
    subj_honesty_data = []
    subj_random_data = []
    subj_all_data = []
    
    for con_id in range(1, 25): 
        con_file = f'extracted_con_{con_id:04d}.nii'
        con_file_path = os.path.join(subject_dir, con_file)

        if not os.path.exists(con_file_path):
            print(f"File {con_file_path} not found! Skipping.")
            continue
        
        img = nib.load(con_file_path)
        con_data = img.get_fdata()

        if con_id in lie_indices:
            subj_lie_data.append(con_data)
        elif con_id in honesty_indices:
            subj_honesty_data.append(con_data)
        elif con_id in random_indices:
            subj_random_data.append(con_data)
        
        subj_all_data.append(con_data)


    elie = np.mean(subj_lie_data, axis=0)
    ehon = np.mean(subj_honesty_data, axis=0)
    erand = np.mean(subj_random_data, axis=0)
    eall = np.mean(subj_all_data, axis=0)

    all_subjects_elie.append(elie.flatten())
    all_subjects_ehon.append(ehon.flatten())
    all_subjects_erand.append(erand.flatten())
    all_subjects_eall.append(eall.flatten())

df_elie = pd.DataFrame(all_subjects_elie, index=subject_ids)
df_ehon = pd.DataFrame(all_subjects_ehon, index=subject_ids)
df_erand = pd.DataFrame(all_subjects_erand, index=subject_ids)
df_eall = pd.DataFrame(all_subjects_eall, index=subject_ids)

with pd.ExcelWriter(output_excel) as writer:
    df_elie.to_excel(writer, sheet_name='elie')
    df_ehon.to_excel(writer, sheet_name='ehon')
    df_erand.to_excel(writer, sheet_name='erand')
    df_eall.to_excel(writer, sheet_name='eall')

print(f"Average activation values saved to {output_excel}")
```
